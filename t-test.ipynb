```python
import numpy as np
from scipy import stats

# N = 20  # episodes per method

def welch_ttest(m1, s1, m2, s2, n1=20, n2=10):
    se  = np.sqrt(s1**2/n1 + s2**2/n2)
    t   = (m1 - m2) / se
    df  = (s1**2/n1 + s2**2/n2)**2 / ((s1**2/n1)**2/(n1-1) + (s2**2/n2)**2/(n2-1))
    p_value   = 2 * stats.t.sf(abs(t), df)
    return t, df, p_value

def report(mifomo, baselines, significance_level=0.05):
    for metric_id, metric in enumerate(["OA", "AA", "KC"]):
        m1, s1 = mifomo[metric_id]
        print(f"\n->{metric}: MIFOMO ({m1} ± {s1}) vs baselines")
        for name, vals in baselines.items():
            m2, s2 = vals[metric_id]
            t, df, p = welch_ttest(m1, s1, m2, s2)
            # print(f"{name:10s} \t t={t:6.2f} \t diff={m1 - m2:.2f} \t df={df:5.1f} \t p={p:.4g} \t sig.={p < significance_level}")
            print(f"{name:10s} \t t={t:6.2f} \t diff={m1 - m2:.2f} \t p={p:.4g} \t sig.={p < significance_level}")

def report_latex(mifomo, baselines, significance_level=0.05):
    for name, vals in baselines.items():
        res = ""
        
        for metric_id, metric in enumerate(["OA", "AA", "KC"]):
            m1, s1 = mifomo[metric_id]
            m2, s2 = vals[metric_id]
            t, df, p_value = welch_ttest(m1, s1, m2, s2)
            res += " & \\val{" + f"{m2}" + "}{" + f"{s2}" + "} & "
            res += f"{m1 - m2:.2f} & {p_value:.4g} & " 
            res += "\\cmark" if p_value < significance_level else "\\xmark "
        print(f"{name:10s} {res} \\\\")
```
